**WAP ask the user enter gender either male or female**
- if gender equal to male
- ask enter age
- if age is greater than 60 print senior citizen
- if age between 30 to 60 print middle age
- if age between 20 to 30 print young boy
- if age between 13 to 20 print teen age boy
- otherwise;print(boy)
- else:
if gender equal to female
- ask enter age
- if age is greater than 60 print senior citizen
- if age between 30 to 60 print middle age
- if age between 20 to 30 print young lady
- if age between 13 to 20 print teen age girl
- otherwise;print(girl)

```python
Gender=input('Enter your sex')
if Gender=='male':
    age=eval(input('Enter your age'))
    if age>60:
        print('Senior citizen')
    if age>30:
        print('Middle age men')
    if age>20:
        print('Young boy')
    if age>13:
        print('Teen_age boy')
elif Gender=='female':
    age=eval(input('Enter your age'))
    if age>60:
        print('Senior citizen')
    if age>30:
        print('Middle age lady')
    if age>20:
        print('Young lady')
    if age>13:
        print('Teen_age lady')
else:
    print('Enter valid gender')
```
