**Take 5  questions in list1**
- take correspondence 5 answers in another list2
- iterate through list1
- every queestion should print
- answer =input('tell the answer')
- answer will check in list2
- if it is there:marks=marksa+1
- else:no marks
- how many answers you made ,how many answers you got

```python
list1=['Who is PM of Nepal?','What is capital of Nepal?','Who is thug of our house?','Who is don of our house?','Who is intelligent,genious of our house?']
list2=['Prachanda','Kathmandu','Suti','Mom','Sushankhya']
right_marks_count=0
wrong_marks_count=0
for i in range(len(list1)):
    print(list1[i])
    answer=input('Enter your answer')
    if answer==list2[i]:
        print('Your answer is correct')
        right_marks_count=right_marks_count+1
    else:
        print('Your answers are wrong')
        wrong_marks_count=wrong_marks_count+1

print('The number of answers you made right:',right_marks_count)
print('The number of answers you made wrong:',wrong_marks_count)
```
