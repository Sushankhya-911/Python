**WAP ask the useer enter distance in km**
- if distance is more than  km:print charge is 100 rs
- if distance is between 7 to 100km:print charge is 70rs
- if distance is between 4 to 7km:print charge is 40rs
- otherwise print charge is 30rs

```python
distance=eval(input('Enter a number:'))
if distance>=10:
    print('The charge is 100 rs')

elif distance>=7:
    print('The charge is 70rs')

elif distance>=4:
    print('The charge is 40rs')

else:
    print('The charge is 30rs')
```
