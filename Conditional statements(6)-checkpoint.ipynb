```python
import random
random.randint(1,200)
value=random.randint(1,200)
if value>=10 and value<=99:
    print('I won')

else:
    print('I lost')
```

```python
try:
    import random
    random.randint(1,200)
    value=random.randint(1,200)
    if value>=10 and value<=99:
        print('I won')

    else:
        print('I lost')

except Exception as e:
    print('e')
```

**WAP ask the user enter percentage of marks**
- if per>90 print A
- if per between 70to 90 print B
- if per between 50 to 79 print C
- otherwise print D
- if-elif-elif-else

```python
num1=eval(input('enter a number:'))
if num1>90:
    print('A grade')

elif 70<num1<90:
    print('B grade')

elif 50<num1<70:
    print('C grade')

else:
    print('D grade')
    
```
