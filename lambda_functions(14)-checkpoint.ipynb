**Create a function for finding greater number between two numbers**

```python
def greater(n1,n2):
    if n1<n2:
        print('{} is greater'.format(n2))
    else:
        print('{} is greater'.format(n1))

greater(10,20)
```

```python
l1=[]
l2=[]
def greater(n1,n2):
    if n1<n2:
       l1.append(n2)
    else:
       l2.append(n1)

greater(10,20)
```
