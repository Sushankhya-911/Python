```python
n1=20  #global variable
n2=30  # global variable 
def add(n1):
    n2=200
    add=n1+n2
    print('The addition of {} and {} is {}'.format(n1,n2,add))

add(500)
print(n1)   # global variable
```

```python
n1=20  #global variable
n2=30  # global variable 
def add(n1=700):
    n2=200
    add=n1+n2
    print('The addition of {} and {} is {}'.format(n1,n2,add))

add(500)
print(n1)   # global variable
```

**Create four functions**
- addition func
- subtraction
- multiplication
- division
- provide print statements



```python
def add(a,b):
    print(a+b)

def subtract(a,b):
    print(a-b)

def multiply(a,b):
    print(a*b)

def divide(a,b):
    print(a/b)

def main():
    print('If you want to perform addition :Enter (1)')
    print('If you want to perform subtraction :Enter (2)')
    print('If you want to perform multiplication :Enter (3)')
    print('If you want to perform division :Enter (4)')
    Operation=eval(input('Enter the operation code you want to perform:'))
    a=eval(input('Enter the number'))
    b=eval(input('Enter the number'))
    if Operation==1:
            add(a,b)
    elif Operation==2:
            subtract(a,b)
    elif Operation==3:
            multiply(a,b)
    elif Operation==4:
            divide(a,b)
    else:
            print('Provide correct operation code')

main()
    
```
