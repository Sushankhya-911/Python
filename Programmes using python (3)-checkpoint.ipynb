**Write a program ask the user enter a basic_salary**
- Ask the user enter DA amount
- Ask the user enter HRA amount
- Calulate total pay=basic_salary+DA+HRA 
    


```python
basic_salary=eval(input("enter a basic_salary:"))
DA=eval(input("enter a DA amount:"))
HRA=eval(input("enter a HRA amount:"))
total_pay=basic_salary+DA+HRA
print("the total salary is:",total_pay)
```
