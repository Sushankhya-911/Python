
**Ramesh's basic salary is input through the keyboard. His dearness allowance is 40% of basic salary, and house rent allowance is 20% of basic salary. Write a program to calculate his gross salary**

```python
Ramesh_salary=eval(input("enter a number:"))
Allowance=(40*Ramesh_salary)/100
House_rent=(20*Ramesh_salary)/100
Gross_salary=Ramesh_salary-(House_rent+Allowance)
print("Among {} gross salary of Rameshis {}".format (Ramesh_salary,Gross_salary))    
```

**Python program that reads the distance between two cities in kilometers from the keyboard and converts it to meters, feet, inches, and centimeters**

```python
distance_1=eval(input('ENTER DISTANCE:'))
distance_2=eval(input('ENTER DISTANCE:'))
distance=distance_1 - distance_2
print("The distance is {} meters".format(distance))
distance_cm=distance/100
print("The distance is {} centimeters".format(distance_cm))
distance_in=distance*16
print("The distance is {} inches".format(distance_in))
distance_ft=distance*3.24
print("The distance is {} feet".format(distance_ft))
```

**If the marks obtained by a student in five different subjects are input through the keyboard, find out the aggregate marks and percentage marks obtained by the student. Assume that the maximum marks that can be obtained by a student in each subject is 100.**

```python
Math=eval(input('Enter marks:'))
Science=eval(input('Enter marks:'))
English=eval(input('Enter marks:'))
Social=eval(input('Enter marks:'))
Nepali=eval(input('Enter marks:'))
Total_marks=500
Gained_marks=Math+Science+English+Social+Nepali
percentage_marks=((Gained_marks)*100)/500
print("The total marks among {} achieved by student is {}".format(Total_marks,Gained_marks))
print("The total percentage  achieved by student is {}".format(percentage_marks))
                
```

**Temperature of a city in Fahrenheit degrees is input through the keyboard. Write a program to convert this temperature to  Celsius degrees**

**Calculate the area and perimeter of a rectangle, and the area and circumference of a circle, based on the input provided through the keyboard:**

```python
length=eval(input('Enter a length:'))
breadth=eval(input('Enter a breadth:'))
Area=length*breadth
Perimeter=2*(length+breadth)
print('The area of rectangle is {}'.format(Area))
print('The perimeter of rectangle is {}'.format(Perimeter))
radius=eval(input('Enter a radius:'))
import math
math.pi
pi=math.pi
C_area=pi*radius*radius
C_perimeter=2*pi*radius
print('The area of circle is {}'.format(C_area))
print('The perimeter of circle is {}'.format(C_perimeter))
```

**If a five-digit number is input through the keyboard, write a program to reverse the number.**

```python
num=eval(input('Enter a number:'))
num_string=str(num)
reversed_string=num_string[::-1]
reversed_number=int(reversed_string)
print('The reversed number:',reversed_number)
```

**The Python code to calculate the sum of the first and last digits of a four-digit number:**

```python
num=eval(input('Enter a number:'))
num_string=str(num)
first=int(num_string[0])
last=int(num_string[-1])
add=first+last
print('Sum of  number',add)
```

**NOTE:As I have observed that whenever we do some operation leading to single digit:such like reverse,add**
- Integer data type cant be used because integer are considered as a single type(1234)==it is considered as one variable.
- Whereas when we use string it is set as a list that is (1234)==considered as 4 element list.
- Where 1st one is denoted as [0] and [-1] as last.
