**Suppose that a Firm pays its employee at the rate of $12 per hour. An employee has worked for 37 
hours. How much should the firm pay the employee? (Do normal way directly takes the input)**

```python
Rate=12
working_hours=37
Payable_amount=37*12
print('The payable amount at rate of {} and working hours {} is {}'.format(Rate,working_hours,Payable_amount))
```

**How about prompting the user for the number of hours and using the input value to compute the 
total pay? (Above problem use eval concept and sep concept)**

```python
number_of_hours=eval(input('Enter the number of hours:'))
Rate=eval(input('Enter the number of rate:'))
total_pay=number_of_hours*Rate
print('Total pay:',total_pay,sep='$')
```

- Display the grades of students using the following table:

```python
Quiz_1=eval(input('Enter the number:'))
Quiz_2=eval(input('Enter the number:'))
class_test=eval(input('Enter the number:'))
assignment=eval(input('Enter the number:'))
project=eval(input('Enter the number:'))

total_marks=Quiz_1+Quiz_2+class_test+assignment+project
percentage=(total_marks/500)*100
print('Total_marks:',total_marks)
print('Percentage:',percentage)
if total_marks>=60:
    print('First class')
elif total_marks>=40:
    print('Second class')
elif total_marks<40:
    print('Fail')
else:
    print('Not Valid')
```

**Rewrite the pay program (refer problem 2) using try and expect so that the program handles non- 
numeric input gracefully by printing a message and exiting the program. The following shows two 
executions of the program:**
- Enter hours: 20
- Enter rate: nine
- Error, please enter numeric 


```python
try:
    number_of_hours=eval(input('Enter the number of hours:'))
    Rate=eval(input('Enter the number of rate:'))
    total_pay=number_of_hours*Rate
    print('Total pay:',total_pay,sep='$')

except Exception as e:
    print('Please enter numeric value')
    print(e)
```
